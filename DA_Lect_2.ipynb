```python
# Lecture 2 - Exaple 1  - Descriptive Analysis by Hand
# Store the sales data in a list
sales = [420, 455, 610, 300, 175]

# Store the region names in another list
regions = ["North", "South", "East", "West", "Central"]

# Find the total sales
total = sum(sales)

# Find the average sales
average = total / len(sales)

# Find the lowest sales value
lowest_sales = min(sales)

# Find the index of the lowest sales
index = sales.index(lowest_sales)

# Get the name of the lowest performing region
lowest_region = regions[index]

# Display the results
print("Total sales:", total)
print("Average sales:", average)
print("Lowest performing region:", lowest_region)

# Check the output below
```

```python
# Lecture 2 - Exaple 2  - A Minimal Chart Using Matplotlib (Preview)
# Import the matplotlib library for plotting graphs (using matplot for graph)
import matplotlib.pyplot as plt

# Store the region names in a list
regions = ["North", "South", "East", "West", "Central"]

# Store the sales values in a list
sales = [420, 455, 610, 300, 175]

# Create a bar chart
plt.bar(regions, sales)

# Add a title to the graph
plt.title("Regional Sales")

# Add a label to the Y-axis
plt.ylabel("Sales (₹ thousands)")

# Display the graph
plt.show()
# Check the output below
```

```python
# Lecture 2 - Exaple 3  -A Simple Diagnostic Check
# Store the region names
regions = ["North", "South", "East", "West", "Central"]

# Store the sales values
sales = [420, 455, 610, 300, 175]

# Find the sales of all regions except Central
other_sales = []

for i in range(len(regions)):
    if regions[i] != "Central":
        other_sales.append(sales[i])

# Calculate the average sales of the other regions
avg_others = sum(other_sales) / len(other_sales)

# Find the sales of the Central region
central_sales = sales[regions.index("Central")]

# Calculate the difference
gap = avg_others - central_sales

# Display the result
print("Central trails the other regions' average by", round(gap, 1), "(₹ thousands)")
# Check the output below
```
