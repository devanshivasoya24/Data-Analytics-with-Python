```python
# Lecture 3 - Exaple 1  - Checking Data Types
# Integer value
a = 10

# Decimal (float) value
b = 10.5

# Text (string) value
c = "Data Analysis"

# True/False value (Boolean)
d = True

# Print the data type of each variable
print(type(a))
print(type(b))
print(type(c))
print(type(d))
```

```python
# Lecture 3 - Exaple 2  - Type Casting
# Store 100 as a string
x = "100"

# Convert string into integer and add 50
y = int(x) + 50

# Print the result and its data type
print(y)
print(type(y))
```

```python
# Lecture 3 - Exaple 3  - Lists and Membership Testing
# Create a list of fruits
fruits = ["apple", "banana", "mango"]

# Check if banana is present in the list
print("banana" in fruits)
```

```python
# Lecture 3 - Exaple 4  - Tuples (Immutable Sequences)
# Create a tuple
# A tuple cannot be changed after it is created
t = (1, 2, 3)

# Print the tuple and its data type
print(t)
print(type(t))

# Try to change the first value
try:
    t[0] = 99

# This error occurs because tuples cannot be changed
except TypeError as e:
    print("Error:", e)
```

```python
# Lecture 3 - Exaple 5  - Sets Automatically Remove Duplicates
# Create a set
# Duplicate values are automatically removed
s = {1, 2, 2, 3}

# Print the set
print(s)
```

```python
# Lecture 3 - Exaple 6  - Dictionaries for Named Fields
# Create a dictionary with keys and values
record = {"a": 1, "b": 2}

# Get the value of key "a"
print(record["a"])
```

```python
# Lecture 3 - Exaple 7  - Complex Numbers
# Create a complex number
z = 3 + 4j

# Print the real part
print("Real part:", z.real)

# Print the imaginary part
print("Imaginary part:", z.imag)

# Print the absolute value
print("Absolute value:", abs(z))
```
